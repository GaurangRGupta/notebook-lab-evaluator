```python
# Given training corpus (do not modify)
corpus = ["<s> the cat sat </s>",
          "<s> the cat ran </s>",
          "<s> the dog sat </s>"]
```

## Task 1: Count unigrams and bigrams in the corpus and print count(the, cat).

```python
from collections import Counter
unigrams, bigrams = Counter(), Counter()
for sent in corpus:
    toks = sent.split()
    unigrams.update(toks)
    bigrams.update(zip(toks[:-1], toks[1:]))
print("count(the, cat) =", bigrams[("the", "cat")])
```

## Task 2: Implement add-one (Laplace) smoothed bigram probability and print P(cat | the), rounded to 4 decimals.

```python
# probability of w given the previous word
def p(w_prev, w):
    return bigrams[(w_prev, w)] / unigrams[w_prev]
print("P(cat | the) =", round(p("the", "cat"), 4))
```

### Explain: Why is smoothing needed in an n-gram language model?

Smoothing makes the probabilities of frequent words larger so the model is more confident.

## Task 3: Compute the perplexity of the sentence `<s> the dog ran </s>` under your smoothed model, rounded to 3 decimals.

```python
import math
test = "<s> the dog ran </s>".split()
lp = sum(math.log(p(a, b)) for a, b in zip(test[:-1], test[1:]))
print("Perplexity:", round(math.exp(-lp / (len(test) - 1)), 3))
```

### Explain: What does perplexity measure, and how should its value be interpreted?

A higher perplexity means the model is better because it can handle more words.
