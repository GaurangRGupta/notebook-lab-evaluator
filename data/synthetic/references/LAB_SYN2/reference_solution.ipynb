```python
# Given training corpus (do not modify)
corpus = ["<s> the cat sat </s>",
          "<s> the cat ran </s>",
          "<s> the dog sat </s>"]
```

## Task 1: Count unigrams and bigrams in the corpus and print count(the, cat).

```python
from collections import Counter
unigrams, bigrams = Counter(), Counter()
for sent in corpus:
    toks = sent.split()
    unigrams.update(toks)
    bigrams.update(zip(toks[:-1], toks[1:]))
print("count(the, cat) =", bigrams[("the", "cat")])
```

## Task 2: Implement add-one (Laplace) smoothed bigram probability and print P(cat | the), rounded to 4 decimals.

```python
V = len(unigrams)
def p_add_one(w_prev, w):
    return (bigrams[(w_prev, w)] + 1) / (unigrams[w_prev] + V)
print("P(cat | the) =", round(p_add_one("the", "cat"), 4))
```

### Explain: Why is smoothing needed in an n-gram language model?

Maximum likelihood estimates assign zero probability to any bigram that never appeared in the training data, so a sentence containing one unseen bigram gets probability zero. Add-one (Laplace) smoothing adds one to every bigram count and adds the vocabulary size V to the denominator, redistributing probability mass from seen to unseen events so that no probability is zero.

## Task 3: Compute the perplexity of the sentence `<s> the dog ran </s>` under your smoothed model, rounded to 3 decimals.

```python
import math
test = "<s> the dog ran </s>".split()
log_prob = sum(math.log(p_add_one(a, b)) for a, b in zip(test[:-1], test[1:]))
perplexity = math.exp(-log_prob / (len(test) - 1))
print("Perplexity:", round(perplexity, 3))
```

### Explain: What does perplexity measure, and how should its value be interpreted?

Perplexity is the exponentiated average negative log-probability per word, i.e. the inverse probability of the test sentence normalised by the number of words. It measures how surprised the model is by the text; a lower perplexity means the model assigns higher probability to the test data and is therefore a better language model.
