```python
# Given training corpus (do not modify)
corpus = ["<s> the cat sat </s>",
          "<s> the cat ran </s>",
          "<s> the dog sat </s>"]
```

## Task 1: Count unigrams and bigrams in the corpus and print count(the, cat).

```python
from collections import Counter

def count_ngrams(sentences):
    """Return unigram and bigram Counters for a list of tokenised sentences."""
    uni, bi = Counter(), Counter()
    for sent in sentences:
        toks = sent.split()
        uni.update(toks)
        # consecutive token pairs are the bigrams
        bi.update(zip(toks[:-1], toks[1:]))
    return uni, bi

unigrams, bigrams = count_ngrams(corpus)
print("count(the, cat) =", bigrams[("the", "cat")])
```

## Task 2: Implement add-one (Laplace) smoothed bigram probability and print P(cat | the), rounded to 4 decimals.

```python
V = len(unigrams)  # vocabulary size, including <s> and </s>

def p_add_one(w_prev, w):
    """Laplace-smoothed P(w | w_prev)."""
    return (bigrams[(w_prev, w)] + 1) / (unigrams[w_prev] + V)

print("P(cat | the) =", round(p_add_one("the", "cat"), 4))
```

### Explain: Why is smoothing needed in an n-gram language model?

With maximum likelihood estimation, any bigram that was never seen in training gets zero probability, so a whole test sentence containing it would get probability zero. Add-one (Laplace) smoothing adds one to every count and adds the vocabulary size V to the denominator, which redistributes a little probability mass to unseen bigrams.

## Task 3: Compute the perplexity of the sentence `<s> the dog ran </s>` under your smoothed model, rounded to 3 decimals.

```python
import math

def perplexity(sentence):
    """Perplexity of a sentence under the add-one bigram model."""
    toks = sentence.split()
    log_p = sum(math.log(p_add_one(a, b)) for a, b in zip(toks[:-1], toks[1:]))
    return math.exp(-log_p / (len(toks) - 1))

print("Perplexity:", round(perplexity("<s> the dog ran </s>"), 3))
```

### Explain: What does perplexity measure, and how should its value be interpreted?

Perplexity is the inverse probability of the test sentence normalised by the number of words (the exponent of the average negative log-probability per word). It tells us how surprised the model is; a lower perplexity means a better language model.
