```python
# Given corpus (do not modify)
docs = ["the cat sat on the mat",
        "the dog sat on the log",
        "cats and dogs are pets"]
```

## Task 1: Tokenise each document (lowercase, whitespace split) and print the vocabulary size.

```python
def tokenize(doc):
    """Lowercase a document and split it on whitespace."""
    return doc.lower().split()

# the vocabulary is the set of unique tokens across all documents
tokens = [tokenize(d) for d in docs]
vocab = sorted({t for doc in tokens for t in doc})
print("Vocabulary size:", len(vocab))
```

## Task 2: Build a TF-IDF matrix with scikit-learn's `TfidfVectorizer` and print its shape.

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# fit the vectorizer on the corpus; rows = documents, columns = vocabulary terms
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(docs)
print("TF-IDF matrix shape:", X.shape)
```

### Explain: What does TF-IDF measure, and why does it down-weight words such as 'the'?

TF-IDF combines **term frequency** (how often a word occurs in a document) with **inverse document frequency**, the log of the total number of documents over the number of documents containing the word. A word like *the* occurs in almost every document, so its IDF is close to zero and it is down-weighted; rare, distinctive words such as *pets* get a higher weight and therefore dominate the document vector.

## Task 3: Compute the cosine similarity between document 0 and document 1, rounded to 4 decimals.

```python
from sklearn.metrics.pairwise import cosine_similarity

def doc_similarity(i, j):
    """Cosine similarity between rows i and j of the TF-IDF matrix."""
    return cosine_similarity(X[i], X[j])[0, 0]

print("Cosine similarity:", round(doc_similarity(0, 1), 4))
```

### Explain: What does cosine similarity measure, and why is it preferred over Euclidean distance for text?

Cosine similarity is the cosine of the angle between two document vectors: the dot product divided by the product of their norms. Because it only looks at the direction of the vectors, a long and a short document about the same topic can still be similar. Euclidean distance depends on magnitude, so document length would distort it.
