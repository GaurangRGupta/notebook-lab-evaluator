```python
# Given corpus (do not modify)
docs = ["the cat sat on the mat",
        "the dog sat on the log",
        "cats and dogs are pets"]
```

## Task 1: Tokenise each document (lowercase, whitespace split) and print the vocabulary size.

```python
tokens = [d.lower().split() for d in docs]
vocab = sorted({t for doc in tokens for t in doc})
print("Vocabulary size:", len(vocab))
```

## Task 2: Build a TF-IDF matrix with scikit-learn's `TfidfVectorizer` and print its shape.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(docs)
print("TF-IDF matrix shape:", X.shape)
```

### Explain: What does TF-IDF measure, and why does it down-weight words such as 'the'?

TF-IDF (term frequency-inverse document frequency) weights a word by how often it occurs in a document (term frequency) multiplied by its inverse document frequency, the log of the number of documents divided by the number of documents that contain the word. Words such as 'the' appear in most documents, so their IDF is low and they are down-weighted, while rare, discriminative words receive a higher weight.

## Task 3: Compute the cosine similarity between document 0 and document 1, rounded to 4 decimals.

```python
from sklearn.metrics.pairwise import cosine_similarity
sim = cosine_similarity(X[0], X[1])[0, 0]
print("Cosine similarity:", round(sim, 4))
```

### Explain: What does cosine similarity measure, and why is it preferred over Euclidean distance for text?

Cosine similarity is the cosine of the angle between two vectors: their dot product divided by the product of their norms (magnitudes). It depends only on the direction of the vectors, so it is insensitive to document length, whereas Euclidean distance grows with vector magnitude and would make two documents on the same topic but of different lengths look far apart.
