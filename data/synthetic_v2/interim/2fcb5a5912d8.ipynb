```python
# Given training corpus (do not modify)
corpus = ["<s> the cat sat </s>",
          "<s> the cat ran </s>",
          "<s> the dog sat </s>"]
```

## Task 1: Count unigrams and bigrams in the corpus and print count(the, cat).

```python
counts = {}
for sent in corpus:
    toks = sent.split()
    for a, b in zip(toks, toks[1:]):
        counts[(a, b)] = counts.get((a, b), 0) + 1
print("count(the, cat) =", counts[("the", "cat")])
```

## Task 2: Implement add-one (Laplace) smoothed bigram probability and print P(cat | the), rounded to 4 decimals.

```python
from collections import Counter
U, B = Counter(), Counter()
for s in corpus:
    t = s.split()
    U.update(t)
    B.update(zip(t[:-1], t[1:]))
V = len(U)
p = (B[("the", "cat")] + 1 / (U["the"] + V)
print("P(cat | the) =", round(p, 4))
```

### Explain: Why is smoothing needed in an n-gram language model?

Add-one smoothing adds one to every bigram count.

## Task 3: Compute the perplexity of the sentence `<s> the dog ran </s>` under your smoothed model, rounded to 3 decimals.

```python
from collections import Counter
U, B = Counter(), Counter()
for s in corpus:
    t = s.split()
    U.update(t)
    B.update(zip(t[:-1], t[1:]))
import math
V = len(U)
test = "<s> the dog ran </s>".split()
lp = sum(math.log((B[(a, b)] + 1) / (U[a] + V)) for a, b in zip(test[:-1], test[1:])
print("Perplexity:", round(math.exp(-lp / (len(test) - 1)), 3))
```

### Explain: What does perplexity measure, and how should its value be interpreted?

Perplexity is the inverse probability of the test sentence normalised by the number of words, i.e. the exponent of the average negative log-probability per word. A lower perplexity means the model assigns higher probability to the test data and is a better language model.
