```python
# Given corpus (do not modify)
docs = ["the cat sat on the mat",
        "the dog sat on the log",
        "cats and dogs are pets"]
```

## Task 1: Tokenise each document (lowercase, whitespace split) and print the vocabulary size.

```python
def tokenize(text):
    """Lowercase a text and split it on whitespace."""
    return text.lower().split()
```

```python
# build the vocabulary from lowercase whitespace tokens
tokens = [tokenize(d) for d in docs]
# unique tokens across all documents
vocab = sorted({t for doc in tokens for t in doc})
print("Vocabulary size:", len(vocab))
```

## Task 2: Build a TF-IDF matrix with scikit-learn's `TfidfVectorizer` and print its shape.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
# fit TF-IDF on the corpus: rows are documents, columns are terms
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(docs)
print("TF-IDF matrix shape:", X.shape)
```

### Explain: What does TF-IDF measure, and why does it down-weight words such as 'the'?

TF-IDF multiplies the term frequency of a word in a document by its inverse document frequency, the log of the number of documents divided by the number of documents containing the word. Words like 'the' that appear in most documents get a low IDF and are down-weighted, while rare words get a higher weight.

## Task 3: Compute the cosine similarity between document 0 and document 1, rounded to 4 decimals.

```python
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.feature_extraction.text import TfidfVectorizer
M = TfidfVectorizer().fit_transform(docs)
print("Cosine similarity:" round(cosine_similarity(M[0], M[1])[0, 0], 4))
```

### Explain: What does cosine similarity measure, and why is it preferred over Euclidean distance for text?

This is a standard way of comparing documents that works well in practice.
