```python
# Given corpus (do not modify)
docs = ["the cat sat on the mat",
        "the dog sat on the log",
        "cats and dogs are pets"]
```

## Task 1: Tokenise each document (lowercase, whitespace split) and print the vocabulary size.

```python
def tokenize(text):
    """Lowercase a text and split it on whitespace."""
    return text.lower().split()
```

```python
# count the words
words = []
for d in docs:
    words.extend(tokenize(d))
print("Vocabulary size:", len(words))
```

## Task 2: Build a TF-IDF matrix with scikit-learn's `TfidfVectorizer` and print its shape.

```python
from sklearn.feature_extraction.text import TfidfVectorizer as TV
# document-term matrix with tf-idf weights
X = TV().fit_transform(docs)
X.shape
```

### Explain: What does TF-IDF measure, and why does it down-weight words such as 'the'?

Each word is scored by how many times it shows up in a text, scaled down when that word turns up in lots of other texts. So filler words found everywhere end up almost worthless, while words that are specific to a few texts stand out.

## Task 3: Compute the cosine similarity between document 0 and document 1, rounded to 4 decimals.

```python
from sklearn.feature_extraction.text import CountVectorizer
import numpy as np
# similarity as a dot product of counts
C = CountVectorizer().fit_transform(docs).toarray()
print("Cosine similarity:", np.dot(C[0], C[1]))
```

### Explain: What does cosine similarity measure, and why is it preferred over Euclidean distance for text?

Cosine similarity tells us how similar two documents are. It is better than other measures.
