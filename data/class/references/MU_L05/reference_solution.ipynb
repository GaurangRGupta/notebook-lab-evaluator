## Task 1

Load train.csv, show its shape and missing values, and count survivors and non-survivors.

```python
import pandas as pd
df = pd.read_csv("train.csv")
print(df.shape)
print(df.isnull().sum())
print(df["Survived"].value_counts())
```

## Task 2

Plot count plots / bar charts for Sex, Pclass, Parch and Embarked.

```python
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
for col in ["Sex", "Pclass", "Parch", "Embarked"]:
    df[col].value_counts().plot(kind="bar", title=col); plt.show()
print(df["Pclass"].value_counts())
```

## Task 3

Tabulate survival by passenger class and by sex, and the survival rate of each group.

```python
print(pd.crosstab(df["Pclass"], df["Survived"]))
print(pd.crosstab(df["Sex"], df["Survived"]))
print(df.groupby("Sex")["Survived"].mean().round(4))
print(df.groupby("Pclass")["Survived"].mean().round(4))
```

**Explain:** What do the tables say about who was more likely to survive?

Women survived far more often than men (about 74% against 19%), and first class passengers survived more often than third class passengers (about 63% against 24%). Survival depends on sex and class, so these are conditional probabilities that differ from the overall rate of 38%.
