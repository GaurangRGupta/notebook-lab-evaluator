## Task 1

For data = [1, 3, 4, 5, 7, 9, 2] compute the mean by hand (sum / n) and with the statistics module.

```python
import statistics
data = [1, 3, 4, 5, 7, 9, 2]
mean_manual = sum(data) / len(data)
print("manual mean:", round(mean_manual, 4))
print("statistics.mean:", round(statistics.mean(data), 4))
```

## Task 2

Plot a histogram of data, and a histogram of 200 values drawn from a normal distribution with mean 100 and standard deviation 10.

```python
import numpy as np
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
plt.hist(data, bins=5); plt.title("data"); plt.show()
sample = np.random.default_rng(0).normal(100, 10, 200)
plt.hist(sample, bins=20); plt.title("normal(100, 10), n=200"); plt.show()
print("sample mean:", round(float(sample.mean()), 2), "sample sd:", round(float(sample.std(ddof=1)), 2))
```

**Explain:** Describe the shape of the normal histogram and how it relates to the mean and standard deviation.

The normal histogram is roughly bell shaped and symmetric around the mean of 100. Most values lie within one standard deviation (90 to 110) and almost all within three (70 to 130). With only 200 values the bars are uneven.

## Task 3

Compute the median of data by sorting (odd and even n), and check with statistics.median.

```python
s = sorted(data); n = len(s)
median_manual = s[n // 2] if n % 2 else (s[n // 2 - 1] + s[n // 2]) / 2
print("manual median:", median_manual)
print("statistics.median:", statistics.median(data))
```

## Task 4

Find the mode of [1, 3, 4, 5, 7, 9, 1, 7]. It has two modes.

```python
from collections import Counter
d2 = [1, 3, 4, 5, 7, 9, 1, 7]
counts = Counter(d2); top = max(counts.values())
print("modes:", sorted(k for k, v in counts.items() if v == top))
print("statistics.multimode:", statistics.multimode(d2))
```

**Explain:** Why does statistics.mode return a single value here, and why is multimode the better answer?

Both 1 and 7 occur twice, so the data are bimodal. statistics.mode returns only the first mode it meets (1), so it hides the second one. multimode returns every value with the highest count.
