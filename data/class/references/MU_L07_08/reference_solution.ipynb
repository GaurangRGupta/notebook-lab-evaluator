## Task 1

X ~ Binomial(n=12, p=0.45). Find P(X = 8), P(X > 4) and P(X <= 10).

```python
from scipy.stats import binom, poisson
n, p = 12, 0.45
print("P(X=8)  =", round(binom.pmf(8, n, p), 6))
print("P(X>4)  =", round(1 - binom.cdf(4, n, p), 6))
print("P(X<=10)=", round(binom.cdf(10, n, p), 6))
```

## Task 3

Arrivals follow a Poisson distribution with mean 4.2. Find P(X <= 2), P(X >= 5) and P(X = 8).

```python
lam = 4.2
print("P(X<=2) =", round(poisson.cdf(2, lam), 6))
print("P(X>=5) =", round(1 - poisson.cdf(4, lam), 6))
print("P(X=8)  =", round(poisson.pmf(8, lam), 6))
```

**Explain:** When is the Poisson distribution an appropriate model, and why is P(X >= 5) computed as 1 - P(X <= 4)?

The Poisson distribution models the number of events in a fixed interval when events occur independently at a constant average rate. P(X >= 5) is the complement of P(X <= 4), so it is 1 minus the cumulative distribution function at 4. P(X = 8) needs the probability mass function, not the cdf.
