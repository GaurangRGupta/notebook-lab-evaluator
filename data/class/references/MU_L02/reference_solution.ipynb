## Task 1

Numpy exercises Q1-Q19 (arrays, identity matrix, random numbers, linspace, indexing on mat = np.arange(1,26).reshape(5,5), sum, standard deviation, column sums).

```python
import numpy as np
print(np.zeros(10)); print(np.ones(10) * 5); print(np.ones(10))
print(np.arange(10, 51, 2)); print(np.arange(10, 51))
print(np.eye(3)); print(np.arange(9).reshape(3, 3))
rng = np.random.default_rng(0)
print(rng.random()); print(rng.standard_normal(25))
print(np.linspace(0, 1, 20))
mat = np.arange(1, 26).reshape(5, 5)
print(mat[:3, :2]); print(mat[3, 4]); print(mat[:3, 1]); print(mat[-1]); print(mat[-2:])
print("sum:", mat.sum())
print("std:", round(float(mat.std()), 4))
print("column sums:", mat.sum(axis=0))
```

**Explain:** Explain the difference between np.arange and np.linspace.

np.arange steps from a start value by a fixed step size and excludes the stop value. np.linspace returns a fixed number of evenly spaced points and includes both end points.

## Task 2

Pandas exercises Q1-Q12 on population_by_county.xlsx (head, columns, number of states, most common county names, most populated counties and states, counties above 1 million, names without 'County', percent change 2010-2017).

```python
import pandas as pd
pop = pd.read_excel("population_by_county.xlsx")
print(pop.head()); print(pop.columns.tolist())
print("states:", pop["State"].nunique())
print(pop["State"].unique()[:5])
print(pop["County"].value_counts().head())
print(pop.sort_values("2010Census", ascending=False).head())
print(pop.groupby("State")["2010Census"].sum().sort_values(ascending=False).head())
print("counties over 1 million:", (pop["2010Census"] > 1_000_000).sum())
print("names without 'County':", (~pop["County"].str.contains("County")).sum())
pop["PercentChange"] = 100 * (pop["2017PopEstimate"] - pop["2010Census"]) / pop["2010Census"]
st = pop.groupby("State")[["2010Census", "2017PopEstimate"]].sum()
st["PercentChange"] = 100 * (st["2017PopEstimate"] - st["2010Census"]) / st["2010Census"]
print(st.sort_values("PercentChange", ascending=False).head())
```

**Explain:** Explain how you counted the number of states and why value_counts answers Q6.

nunique counts the distinct values of the State column, so it gives the number of states (51 including the District of Columbia). value_counts counts how often each county name occurs and sorts by frequency, so its head gives the most common names.
