## Task 1

Two groups of n = 20 have means 44 and 28 and standard deviations 30 and 20. Test whether the difference of means exceeds 2 at alpha = 0.05 (one-tailed, pooled variance).

```python
import math
from scipy import stats

def two_sample_t(x1, x2, s1, s2, n1, n2, delta, alpha, tails):
    df = n1 + n2 - 2
    sp2 = ((n1 - 1) * s1 ** 2 + (n2 - 1) * s2 ** 2) / df
    t = (x1 - x2 - delta) / math.sqrt(sp2 * (1 / n1 + 1 / n2))
    if tails == 2:
        crit = stats.t.ppf(1 - alpha / 2, df); reject = abs(t) > crit
    else:
        crit = stats.t.ppf(1 - alpha, df); reject = t > crit
    print(f"t = {t:.4f}, critical = {crit:.4f}, df = {df}")
    print("Reject the null hypothesis" if reject else "Fail to reject the null hypothesis")
    return t

two_sample_t(44, 28, 30, 20, 20, 20, 2, 0.05, 1)
```

## Task 2

Two machines: means 200 and 250, standard deviations 20 and 25, n = 25 each. Test for a difference at alpha = 0.01 (two-tailed).

```python
two_sample_t(200, 250, 20, 25, 25, 25, 0, 0.01, 2)
```

**Explain:** Explain the decision, and why the degrees of freedom are n1 + n2 - 2.

The absolute t statistic (about 7.8) is far above the two-tailed critical value at alpha = 0.01, so we reject the null hypothesis that the machines have equal means. With a pooled variance two means are estimated from the data, so the degrees of freedom are n1 + n2 - 2 = 48.
