## Task 1

Simulate N rolls of one fair die. Print the frequency and relative frequency of each face and compare with 1/6.

```python
import numpy as np
rng = np.random.default_rng(42)
N = 10_000
rolls = rng.integers(1, 7, size=N)
for face in range(1, 7):
    f = int((rolls == face).sum())
    print(face, f, round(f / N, 4), "theoretical", round(1 / 6, 4))
```

## Task 2

Simulate N rolls of two dice. For each sum 2-12 print the empirical probability next to the theoretical one, and plot both.

```python
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
sums = rng.integers(1, 7, size=N) + rng.integers(1, 7, size=N)
emp, theo = [], []
for s in range(2, 13):
    emp.append((sums == s).mean()); theo.append((6 - abs(s - 7)) / 36)
    print(s, round(emp[-1], 4), round(theo[-1], 4))
plt.bar(range(2, 13), emp, alpha=0.6, label="empirical"); plt.plot(range(2, 13), theo, "ro-", label="theoretical")
plt.legend(); plt.show()
```

**Explain:** Explain why 7 is the most likely sum and why the empirical probabilities approach the theoretical ones.

Seven can be made in six of the 36 equally likely outcomes, more than any other sum, so it has probability 6/36. By the law of large numbers the relative frequencies get closer to the theoretical probabilities as the number of rolls grows.
