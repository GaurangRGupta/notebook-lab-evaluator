## Task 1

A machine should fill 12 oz. A sample of n = 10 has mean 11.8 and s = 0.15. Test H0: mu = 12 against a two-tailed alternative at alpha = 0.05.

```python
import math
from scipy import stats

def one_sample_t(xbar, mu, s, n, alpha, tails):
    t = (xbar - mu) / (s / math.sqrt(n))
    df = n - 1
    if tails == 2:
        crit = stats.t.ppf(1 - alpha / 2, df); reject = abs(t) > crit
    else:
        crit = stats.t.ppf(1 - alpha, df); reject = t > crit
    print(f"t = {t:.4f}, critical = {crit:.4f}, df = {df}")
    print("Reject the null hypothesis" if reject else "Fail to reject the null hypothesis")
    return t

one_sample_t(11.8, 12, 0.15, 10, 0.05, 2)
```

## Task 3

Scores [62, 92, 75, 68, 83, 95]. Test whether the mean exceeds 70 at the 90% confidence level (one-tailed).

```python
import numpy as np
scores = np.array([62, 92, 75, 68, 83, 95])
t3 = one_sample_t(scores.mean(), 70, scores.std(ddof=1), len(scores), 0.10, 1)
print(stats.ttest_1samp(scores, 70, alternative="greater"))
```

**Explain:** State the hypotheses, the decision and what it means in context.

H0: the mean score is 70; H1: the mean score is greater than 70. The t statistic is larger than the one-tailed critical value at alpha = 0.10 (equivalently the p-value is below 0.10), so we reject the null hypothesis and conclude the mean score is above 70. The sample uses the sample standard deviation with n - 1 degrees of freedom.
