## Task 1

A rental company charges per day: SUV Rs 1000, Sedan Rs 500, Hatchback Rs 250. Rentals longer than 10 days get a 10% discount. A refundable deposit of Rs 500 is collected up front and deducted from the amount payable at return. Read the car type and number of days and print the bill (the reference fixes the inputs instead of using input()).

```python
RATES = {"SUV": 1000, "SEDAN": 500, "HATCHBACK": 250}
DEPOSIT = 500

def rental_bill(car_type, days):
    gross = RATES[car_type.upper()] * days
    discount = 0.10 * gross if days > 10 else 0.0
    payable = gross - discount - DEPOSIT
    return gross, discount, payable

car_type, days = "SUV", 11   # in class this came from input()
gross, discount, payable = rental_bill(car_type, days)
print(f"Car: {car_type}, days: {days}")
print(f"Gross: {gross:.0f}, discount: {discount:.0f}, payable after deposit: {payable:.0f}")
```

## Task 2

For one month (4 weeks) the SUV is rented on Mon/Wed/Fri, the Sedan on Tue/Thu/Sat and the Hatchback on Sunday. Using functions, print the bill for each car and the total.

```python
schedule = {"SUV": 3 * 4, "SEDAN": 3 * 4, "HATCHBACK": 1 * 4}
total = 0
for car, d in schedule.items():
    g, disc, pay = rental_bill(car, d)
    total += pay
    print(f"{car}: {d} days, gross {g:.0f}, discount {disc:.0f}, payable {pay:.0f}")
print(f"Total payable for the month: {total:.0f}")
```

**Explain:** Explain why the SUV and Sedan bills get the discount but the Hatchback bill does not, and how the deposit is treated.

The SUV and the Sedan are each rented for 12 days, which is more than the 10 day threshold, so a 10% discount is applied to their gross amount. The Hatchback is rented for only 4 days, so no discount applies. The refundable deposit of 500 is deducted from each payable amount. Writing the bill as a function lets the same rule be reused for every car.
