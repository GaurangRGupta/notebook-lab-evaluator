## Task 1

Load kerala.csv. Encode FLOODS as 1/0 and add indicator columns for June and July rainfall above 500 mm.

```python
import pandas as pd
df = pd.read_csv("kerala.csv")
df.columns = df.columns.str.strip()
df["F"] = (df["FLOODS"] == "YES").astype(int)
df["JUN_GT_500"] = (df["JUN"] > 500).astype(int)
df["JUL_GT_500"] = (df["JUL"] > 500).astype(int)
print(df.shape, df["F"].sum())
```

## Task 2

Cross-tabulate floods against June rain above 500 mm. Compute P(F), P(J), P(F and J) and P(F | J).

```python
n = len(df)
print(pd.crosstab(df["F"], df["JUN_GT_500"]))
pF = df["F"].mean(); pJ = df["JUN_GT_500"].mean()
pFJ = ((df["F"] == 1) & (df["JUN_GT_500"] == 1)).mean()
print("P(F) =", round(pF, 4), "P(J) =", round(pJ, 4), "P(F and J) =", round(pFJ, 4))
print("P(F | J) =", round(pFJ / pJ, 4))
```

**Explain:** Does heavy June rain change the chance of a flood year? Compare P(F | J) with P(F).

P(F | J) is the probability of a flood year given that June rain exceeded 500 mm. It is about 0.58, a little higher than the unconditional P(F) of about 0.51, so heavy June rain raises the chance of floods only slightly. The events are not independent, but June rain alone is a weak predictor.

## Task 3

Repeat for July rain above 500 mm, and also compute P(July > 500 | F) using Bayes' rule.

```python
pJul = df["JUL_GT_500"].mean()
pFJul = ((df["F"] == 1) & (df["JUL_GT_500"] == 1)).mean()
print(pd.crosstab(df["F"], df["JUL_GT_500"]))
print("P(F | JUL) =", round(pFJul / pJul, 4))
print("P(JUL | F) =", round(pFJul / pF, 4), "via Bayes:", round((pFJul / pJul) * pJul / pF, 4))
```
