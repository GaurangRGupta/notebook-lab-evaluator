```python
# Given training corpus (do not modify)
corpus = ["<s> the cat sat </s>",
          "<s> the cat ran </s>",
          "<s> the dog sat </s>"]
```

## Task 1: Count unigrams and bigrams in the corpus and print count(the, cat).

```python
def tokens(sentence):
    return sentence.split()
```

```python
from collections import Counter
# unigram and bigram counts
unigrams, bigrams = Counter(), Counter()
for sent in corpus:
    toks = tokens(sent)
    unigrams.update(toks)
    # adjacent token pairs are bigrams
    bigrams.update(zip(toks[:-1], toks[1:]))
print("count(the, cat) =", bigrams[("the", "cat")])
```

## Task 2: Implement add-one (Laplace) smoothed bigram probability and print P(cat | the), rounded to 4 decimals.

```python
from collections import Counter
U, B = Counter(), Counter()
for s in corpus:
    t = tokens(s)
    U.update(t)
    B.update(zip(t[:-1], t[1:]))
# add-one smoothing: (count + 1) / (history count + V)
V = len(U)
def p_add_one(w_prev, w):
    return (B[(w_prev, w)] + 1) / (U[w_prev] + V)
print("P(cat | the) =", round(p_add_one("the", "cat"), 4))
```

### Explain: Why is smoothing needed in an n-gram language model?

Without smoothing, unseen bigrams get zero probability. Laplace smoothing adds one to each count and adds the vocabulary size to the history count, redistributing some probability mass to unseen events.

## Task 3: Compute the perplexity of the sentence `<s> the dog ran </s>` under your smoothed model, rounded to 3 decimals.

```python
from collections import Counter
U, B = Counter(), Counter()
for s in corpus:
    t = tokens(s)
    U.update(t)
    B.update(zip(t[:-1], t[1:]))
V = len(U)
test = "<s> the dog ran </s>".split()
prob = 1.0
for a, b in zip(test[:-1], test[1:]):
    # multiply smoothed bigram probabilities
    prob *= (B[(a, b)] + 1) / (U[a] + V)
print("Perplexity:", round(prob ** (-1 / (len(test) - 1)), 3))
```

### Explain: What does perplexity measure, and how should its value be interpreted?

Perplexity is computed from the probability of the test sentence.
